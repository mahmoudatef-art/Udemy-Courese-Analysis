# Handiling Of Meta Data

```python
df_copy.info()
```

```python
df_copy['price'].value_counts()
```

```python
df_copy['price'].replace({'Free' :0, 'TRUE' :0}, inplace = True)
```

```python
df_copy['price'].value_counts()
```

```python
df_copy['price'] = pd.to_numeric(df_copy['price'], errors = 'coerce')
```

```python
df_copy.info()
```

```python
df_copy['content_duration'].value_counts()
```

```python
df_copy['content_duration'] = df_copy['content_duration'].str.replace('hours', '').astype(float)
```

```python
df_copy['content_duration'] = df_copy['content_duration'].str.replace('hours', '').str.replace('hour', '').astype(float)
```

```python
df_copy['content_duration'] = df_copy[
    'content_duration'
    ].str.replace(
        'hours', ''
    ).str.replace(
        'hour', ''
    ).str.replace(
        'mins', ''
    ).astype(float)
```

```python
df_copy['content_duration'] = df_copy[
    'content_duration'
    ].str.extract(
        '(\d+)'
    ).astype(float)
```

```python
df_copy.info()
```

```python
df_copy['published_timestamp'] = pd.to_datetime(df_copy['published_timestamp'])
```

```python
df_copy['content_duration'].isna().sum()
```

```python
df_copy.isna().sum()
```

```python
df_copy[(df_copy['published_timestamp'] == '3 hours')]
```

```python
df_copy['published_timestamp'] = df_copy['published_timestamp'].astype(str)
```

```python
df_copy = df_copy[~df_copy['published_timestamp'].str.contains('3 hours')]
```

```python
df_copy['published_timestamp'].value_counts()
```

```python
df_copy.info()
```

```python
df_copy['date'] = df_copy['published_timestamp'].dt.date
df_copy['time'] = df_copy['published_timestamp'].dt.time
df_copy['year'] = df_copy['published_timestamp'].dt.year
df_copy['month'] = df_copy['published_timestamp'].dt.month
df_copy['quertare'] = df_copy['published_timestamp'].dt.to_period('Q')
```

```python
df_copy['time'] = df_copy['published_timestamp'].dt.strftime('%I:%M %p')
```

```python
df_copy
```

```python
df_copy['year'].value_counts()
```

```python
df_copy.groupby(df_copy['published_timestamp'].dt.year).size()
```

```python
visuliz = px.line(df_copy.groupby(df_copy['published_timestamp'].dt.year).size(), title = 'Number Of Corusess Per Year')
visuliz.update_xaxes(title_text = 'Year')
visuliz.update_yaxes(title_text = 'Count Of Coursess')
visuliz.show()
                  
```

```python
df_copy.info()
```

```python
df_copy['quirter'] = df_copy['published_timestamp'].dt.to_period('Q').astype(str)
Line_Chart_by_quirter = px.line(df_copy.groupby('quirter').size().reset_index(name = 'count'),
                                x = 'quirter', y = 'count', title = 'Number Of Cursess Per Quirter')
Line_Chart_by_quirter.update_xaxes(title_text = 'Quirter')
Line_Chart_by_quirter.update_yaxes(title_text = 'Nember Of Cursess')
Line_Chart_by_quirter.update_traces(text = df_copy.groupby('quirter').size().reset_index(name = 'count')['count'], textposition = 'top center',
                                    line = dict(color = 'red'))
Line_Chart_by_quirter.show()
```

```python
bar_chart_year_subject = px.bar(df_copy.groupby([df_copy['published_timestamp'].dt.year, 'subject']).size().reset_index(name='count'), 
                                x='published_timestamp', y='count', color='subject', barmode = 'group',
                                title='Number of Courses Published by Subject per Year')
bar_chart_year_subject.update_xaxes(title_text='Year')
bar_chart_year_subject.update_yaxes(title_text='Number of Courses')
bar_chart_year_subject.show()
```

```python
bar_chart_1 = px.bar(df_copy.groupby(['subject', 'level']).size().reset_index(name = 'count'),
                                  x = 'subject', y = 'count', color = 'level', barmode = 'group',
                                  title = 'Count Of Subject By Level')
bar_chart_1.update_xaxes(title_text = 'subject')
bar_chart_1.update_yaxes(title_text = 'Count')
bar_chart_1.show()
                  
                
```

```python
max_num_subscribers_by_suject = df_copy.groupby('subject')['num_subscribers'].max().reset_index()
max_num_subscribers_by_suject_chart = px.bar(max_num_subscribers_by_suject, x = 'subject', y = 'num_subscribers',
                                             title = 'max numbers of subscribers by suject')
max_num_subscribers_by_suject_chart.update_xaxes(title_text = 'subject')
max_num_subscribers_by_suject_chart.update_yaxes(title_text = 'Max Numbers Of Subscribers')
max_num_subscribers_by_suject_chart.update_traces(text = max_num_subscribers_by_suject['num_subscribers'].tolist(),
                                                   textposition = 'outside')
max_num_subscribers_by_suject_chart.show()
```

```python
scaterr_chart = px.scatter(df_copy, x = 'num_reviews', y = 'num_subscribers', 
                           title = 'Relashinships between Reviews And Subscribers')
scaterr_chart.update_xaxes(title_text = 'Number Of Reviews')
scaterr_chart.update_yaxes(title_text = 'Number Of Subscribers')
scaterr_chart.show()
```

```python
bar_chart2 = px.bar(df_copy.groupby([df_copy['published_timestamp'].dt.year, 'is_paid']).size().reset_index(name = 'count'),
                    x = 'published_timestamp', y = 'count', color = 'is_paid', barmode = 'group',
                    title = 'count of paid and free cursess per year')
bar_chart2.update_xaxes(title_text = 'year')
bar_chart2.update_yaxes(title_text = 'number of cursess')
bar_chart2.update_traces(text = df_copy.groupby([df_copy['published_timestamp'].dt.year, 'is_paid']).size().reset_index(name = 'count')['count'],
                         textposition = 'outside'
                        )
bar_chart2.show()
```

```python
bar_chart3 = px.bar(df_copy.groupby('is_paid')['num_reviews'].sum().reset_index(),
                    x = 'is_paid', y = 'num_reviews', 
                    title = 'Sum of reviews by by cursess type')
bar_chart3.update_xaxes(title_text = 'cursess type')
bar_chart3.update_yaxes(title_text = 'sum of reviews')
bar_chart3.update_traces(text = df_copy.groupby('is_paid')['num_reviews'].sum().reset_index()['num_reviews'], textposition = 'outside',
                         marker_color = ['red', 'purple'])
bar_chart3.show()
```

```python
bar_chart4 = px.bar(df_copy.groupby('is_paid')['num_subscribers'].sum().reset_index(),
                    x = 'is_paid', y = 'num_subscribers', 
                    title = 'sum of number subscribers of cursess type')
bar_chart4.update_xaxes(title_text = 'sum of subscribers')
bar_chart4.update_yaxes(title_text = 'cursee type')
bar_chart4.update_traces(text = df_copy.groupby('is_paid')['num_subscribers'].sum().reset_index()['num_subscribers'], 
                         textposition = 'outside', marker_color = ['red', 'purple'])
bar_chart4.show()
```
